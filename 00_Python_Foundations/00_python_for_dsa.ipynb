## 1. Lists — Your Bread and Butter

In DSA, Python `list` = **array**. This is the most common data structure you'll use.

````python
# Creating lists
nums = [10, 20, 30, 40, 50]

# Access by index (0-based)
print(f"First element:  nums[0]  = {nums[0]}")
print(f"Last element:   nums[-1] = {nums[-1]}")
print(f"Length:          len(nums) = {len(nums)}")
````

````python
# Slicing — you probably know this from pandas!
print(f"First 3:    {nums[:3]}")
print(f"Last 2:     {nums[-2:]}")
print(f"Reversed:   {nums[::-1]}")
print(f"Every 2nd:  {nums[::2]}")
````

````python
# Common list operations and their TIME COMPLEXITY (we'll learn Big-O next)
# Think of it as: how many steps does the computer take?

nums = [10, 20, 30, 40, 50]

# Append to end — FAST (1 step)
nums.append(60)
print(f"After append(60):    {nums}")

# Insert at position — SLOW (shifts everything after it)
nums.insert(0, 5)
print(f"After insert(0, 5):  {nums}")

# Pop from end — FAST (1 step)
nums.pop()
print(f"After pop():         {nums}")

# Pop from front — SLOW (shifts everything)
nums.pop(0)
print(f"After pop(0):        {nums}")

# Check if element exists — SLOW (scans entire list)
print(f"30 in nums? {30 in nums}")
````

### Key Takeaway

```
Operation          | Speed     | Why?
-------------------+-----------+---------------------------
nums[i]            | Fast O(1) | Direct memory access
nums.append(x)     | Fast O(1) | Add to end
nums.pop()         | Fast O(1) | Remove from end
nums.insert(0, x)  | Slow O(n) | Must shift all elements
nums.pop(0)        | Slow O(n) | Must shift all elements
x in nums          | Slow O(n) | Must scan entire list
```

## 2. Strings — Immutable Sequences

Strings behave like lists of characters, but you **cannot modify them in place**.

````python
s = "hello world"

# Indexing and slicing — same as lists
print(f"s[0] = '{s[0]}'")
print(f"s[:5] = '{s[:5]}'")
print(f"s[::-1] = '{s[::-1]}'")

# Iterating character by character
for char in s:
    print(char, end=" ")
print()
````

````python
# Strings are IMMUTABLE — you can't do s[0] = 'H'
# Instead, create a new string:

s = "hello"
# s[0] = 'H'  # This would raise TypeError!

# Convert to list, modify, convert back
chars = list(s)
chars[0] = 'H'
s = ''.join(chars)
print(f"Modified string: {s}")

# Useful string methods for DSA
print(f"'abc123'.isalnum()  = {'abc123'.isalnum()}")
print(f"'abc'.isalpha()     = {'abc'.isalpha()}")
print(f"'123'.isdigit()     = {'123'.isdigit()}")
print(f"'Hello'.lower()     = {'Hello'.lower()}")
print(f"ord('a')            = {ord('a')}")
print(f"chr(97)             = {chr(97)}")
````

## 3. Dictionaries (Hash Maps) — Your Best Friend

As a Data Analyst, you've used `groupby()` — dictionaries are the low-level engine behind that.

````python
# Counting occurrences — THE most common DSA pattern
# (Like df['color'].value_counts() but manual)

fruits = ['apple', 'banana', 'apple', 'cherry', 'banana', 'apple']

# Method 1: Manual counting
count = {}
for fruit in fruits:
    if fruit in count:
        count[fruit] += 1
    else:
        count[fruit] = 1
print(f"Manual:       {count}")

# Method 2: Using .get() — cleaner
count = {}
for fruit in fruits:
    count[fruit] = count.get(fruit, 0) + 1
print(f"Using .get(): {count}")

# Method 3: Using collections.Counter — cleanest
from collections import Counter
count = Counter(fruits)
print(f"Counter:      {count}")
print(f"Most common:  {count.most_common(2)}")
````

````python
# defaultdict — auto-initializes missing keys
from collections import defaultdict

# Group items (like SQL GROUP BY or pandas groupby)
words = ['eat', 'tea', 'tan', 'ate', 'nat', 'bat']

groups = defaultdict(list)
for word in words:
    key = ''.join(sorted(word))  # sort letters to find anagrams
    groups[key].append(word)

print("Anagram groups:")
for key, group in groups.items():
    print(f"  {key} -> {group}")
````

## 4. Sets — Fast Lookups

When you need to check "have I seen this before?" — use a set, not a list.

````python
# Finding duplicates
nums = [1, 3, 5, 3, 7, 5, 9]

seen = set()
duplicates = set()
for n in nums:
    if n in seen:        # O(1) lookup — FAST!
        duplicates.add(n)
    seen.add(n)

print(f"Duplicates: {duplicates}")

# Set operations (like SQL UNION, INTERSECT, EXCEPT)
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(f"Union (a | b):        {a | b}")
print(f"Intersection (a & b): {a & b}")
print(f"Difference (a - b):   {a - b}")
````

## 5. Essential Python Patterns for DSA

These patterns come up in almost every DSA problem.

````python
# Pattern 1: enumerate() — get index AND value
fruits = ['apple', 'banana', 'cherry']
for i, fruit in enumerate(fruits):
    print(f"  Index {i}: {fruit}")
````

````python
# Pattern 2: zip() — iterate two lists together
names = ['Alice', 'Bob', 'Charlie']
scores = [85, 92, 78]
for name, score in zip(names, scores):
    print(f"  {name}: {score}")
````

````python
# Pattern 3: List comprehension — concise list building
squares = [x**2 for x in range(10)]
print(f"Squares: {squares}")

evens = [x for x in range(20) if x % 2 == 0]
print(f"Evens:   {evens}")
````

````python
# Pattern 4: Sorting with key functions
words = ['banana', 'pie', 'apple', 'kiwi']

print(f"Alphabetical:  {sorted(words)}")
print(f"By length:     {sorted(words, key=len)}")
print(f"By last char:  {sorted(words, key=lambda w: w[-1])}")

# Sorting tuples (like ORDER BY col1, col2 in SQL)
people = [('Alice', 30), ('Bob', 25), ('Charlie', 30)]
print(f"By age, name:  {sorted(people, key=lambda p: (p[1], p[0]))}")
````

````python
# Pattern 5: Two pointers — VERY common in DSA
# Check if a string is a palindrome

def is_palindrome(s: str) -> bool:
    left, right = 0, len(s) - 1
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True

print(f"'racecar' is palindrome? {is_palindrome('racecar')}")
print(f"'hello'   is palindrome? {is_palindrome('hello')}")
print(f"'madam'   is palindrome? {is_palindrome('madam')}")
````

````python
# Pattern 6: Swapping values (no temp variable needed in Python)
a, b = 10, 20
print(f"Before: a={a}, b={b}")
a, b = b, a
print(f"After:  a={a}, b={b}")

# Reverse a list in-place using swaps
nums = [1, 2, 3, 4, 5]
left, right = 0, len(nums) - 1
while left < right:
    nums[left], nums[right] = nums[right], nums[left]
    left += 1
    right -= 1
print(f"Reversed: {nums}")
````

## 6. The `collections` Module — DSA Power Tools

These are allowed in interviews and make your life much easier.

````python
from collections import deque, Counter, defaultdict, OrderedDict

# deque — double-ended queue (fast append/pop from BOTH ends)
dq = deque([1, 2, 3])
dq.appendleft(0)   # O(1) — unlike list.insert(0, x) which is O(n)
dq.append(4)        # O(1)
print(f"deque: {dq}")
dq.popleft()        # O(1) — unlike list.pop(0) which is O(n)
print(f"after popleft: {dq}")
````

````python
# heapq — priority queue (always gives you the smallest element)
import heapq

nums = [5, 1, 8, 3, 2]
heapq.heapify(nums)  # turns list into a min-heap in-place
print(f"Heap: {nums}")
print(f"Smallest: {heapq.heappop(nums)}")
print(f"Next smallest: {heapq.heappop(nums)}")

# Find top-3 largest
data = [10, 4, 5, 8, 6, 11, 26]
print(f"Top 3: {heapq.nlargest(3, data)}")
print(f"Bottom 3: {heapq.nsmallest(3, data)}")
````

## 7. Writing Functions the DSA Way

In DSA problems, you typically write a function that takes input and returns output.  
Here's the standard pattern.

````python
from typing import List, Optional

# Standard DSA function signature with type hints
def two_sum(nums: List[int], target: int) -> List[int]:
    """
    Given a list of numbers and a target, find two numbers that add up to target.
    Return their indices.
    
    Example: nums = [2, 7, 11, 15], target = 9 -> [0, 1]
    Because nums[0] + nums[1] = 2 + 7 = 9
    """
    seen = {}  # value -> index
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i
    return []  # no solution found

# Test it
print(two_sum([2, 7, 11, 15], 9))   # [0, 1]
print(two_sum([3, 2, 4], 6))        # [1, 2]
print(two_sum([3, 3], 6))           # [0, 1]
````

### Congratulations! That `two_sum` function is actually **LeetCode Problem #1** — the most famous DSA problem!

You just solved it using a **hash map** approach. We'll dive deeper into why this works in later notebooks.

## Solutions (try the exercises first!)

````python
# Solution 1
def unique_elements(nums: List[int]) -> List[int]:
    seen = set()
    result = []
    for n in nums:
        if n not in seen:
            result.append(n)
            seen.add(n)
    return result

print(unique_elements([1, 2, 2, 3, 1, 4]))
````

````python
# Solution 2
def first_unique_char(s: str) -> str:
    count = Counter(s)
    for char in s:
        if count[char] == 1:
            return char
    return ""

print(first_unique_char("leetcode"))
print(first_unique_char("aabb"))
````

````python
# Solution 3
def intersect(nums1: List[int], nums2: List[int]) -> List[int]:
    count1 = Counter(nums1)
    result = []
    for n in nums2:
        if count1[n] > 0:
            result.append(n)
            count1[n] -= 1
    return result

print(intersect([1,2,2,1], [2,2]))
print(intersect([4,9,5], [9,4,9,8,4]))
````
