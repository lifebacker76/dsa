# DSA Practice Tracker

Use this notebook to log every problem you solve. Tracking is key to improving.

---

```python
import json
from datetime import datetime
from pathlib import Path

LOG_FILE = Path("problems_solved.json")

def load_log():
    if LOG_FILE.exists():
        return json.loads(LOG_FILE.read_text())
    return []

def save_log(entries):
    LOG_FILE.write_text(json.dumps(entries, indent=2))

def log_problem(name, platform, difficulty, topic, pattern, solved_independently, notes=""):
    """
    Log a problem you solved.
    
    Args:
        name: e.g. "Two Sum"
        platform: "LeetCode", "NeetCode", "GFG", etc.
        difficulty: "Easy", "Medium", "Hard"
        topic: "Arrays", "Linked Lists", "Trees", etc.
        pattern: "Two Pointers", "Sliding Window", "DFS", etc.
        solved_independently: True/False
        notes: any notes for yourself
    """
    entries = load_log()
    entry = {
        "date": datetime.now().strftime("%Y-%m-%d"),
        "name": name,
        "platform": platform,
        "difficulty": difficulty,
        "topic": topic,
        "pattern": pattern,
        "solved_independently": solved_independently,
        "notes": notes
    }
    entries.append(entry)
    save_log(entries)
    print(f"Logged: {name} ({difficulty}) — {topic}/{pattern}")
    print(f"Total problems solved: {len(entries)}")

def show_stats():
    entries = load_log()
    if not entries:
        print("No problems logged yet. Start solving!")
        return
    
    total = len(entries)
    by_diff = {}
    by_topic = {}
    by_pattern = {}
    independent = sum(1 for e in entries if e["solved_independently"])
    
    for e in entries:
        by_diff[e["difficulty"]] = by_diff.get(e["difficulty"], 0) + 1
        by_topic[e["topic"]] = by_topic.get(e["topic"], 0) + 1
        by_pattern[e["pattern"]] = by_pattern.get(e["pattern"], 0) + 1
    
    print(f"{'='*50}")
    print(f"  TOTAL PROBLEMS SOLVED: {total}")
    print(f"  Solved independently:  {independent}/{total} ({independent*100//total}%)")
    print(f"{'='*50}")
    
    print(f"\n  By Difficulty:")
    for diff in ["Easy", "Medium", "Hard"]:
        count = by_diff.get(diff, 0)
        bar = '█' * count
        print(f"    {diff:8s}: {count:3d} {bar}")
    
    print(f"\n  By Topic:")
    for topic, count in sorted(by_topic.items(), key=lambda x: -x[1]):
        bar = '█' * count
        print(f"    {topic:20s}: {count:3d} {bar}")
    
    print(f"\n  By Pattern:")
    for pattern, count in sorted(by_pattern.items(), key=lambda x: -x[1]):
        bar = '█' * count
        print(f"    {pattern:25s}: {count:3d} {bar}")

print("Practice tracker ready! Use log_problem() and show_stats().")
```
