## Pattern 1: Two Pointers

**When to use**: Array is sorted, or you're comparing elements from both ends (palindromes, partitioning, pairs that sum to a target).

**Idea**: Place one pointer at the start, one at the end. Move them toward each other based on a condition.

```
Array:  [ 1,  3,  5,  7,  9, 11 ]    target = 12
          ^                    ^
         left               right

Step 1:  1 + 11 = 12  ✓  Found!

If sum < target → move left  →  (need bigger number)
If sum > target → move right ←  (need smaller number)
```

**Time**: O(n) — each pointer moves at most n times  
**Space**: O(1)

````python
# Problem 1a: Two Sum II — Input Array is Sorted (LC 167)
# Given sorted array, find two numbers that add up to target.
# Return 1-indexed positions.

def two_sum_sorted(nums, target):
    left, right = 0, len(nums) - 1

    while left < right:
        current_sum = nums[left] + nums[right]
        print(f"  left={left} right={right} | nums[{left}]={nums[left]} + nums[{right}]={nums[right]} = {current_sum}", end="")

        if current_sum == target:
            print("  ✓ Found!")
            return [left + 1, right + 1]
        elif current_sum < target:
            print("  < target → move left →")
            left += 1
        else:
            print("  > target → move right ←")
            right -= 1

    return []


print("=== Two Sum II ===")
nums = [2, 7, 11, 15]
target = 9
print(f"Input: {nums}, target={target}")
result = two_sum_sorted(nums, target)
print(f"Result: {result}\n")

nums = [1, 3, 4, 5, 7, 10, 11]
target = 9
print(f"Input: {nums}, target={target}")
result = two_sum_sorted(nums, target)
print(f"Result: {result}")
````

````python
# Problem 1b: Valid Palindrome (LC 125)
# Check if string is palindrome (ignore non-alphanumeric, case-insensitive)

def is_palindrome(s):
    cleaned = [c.lower() for c in s if c.isalnum()]
    left, right = 0, len(cleaned) - 1

    print(f"  Cleaned: {''.join(cleaned)}")

    while left < right:
        print(f"  left={left}('{cleaned[left]}') vs right={right}('{cleaned[right]}')", end="")
        if cleaned[left] != cleaned[right]:
            print("  ✗ Mismatch!")
            return False
        print("  ✓ Match")
        left += 1
        right -= 1

    print("  All matched — palindrome!")
    return True


print("=== Valid Palindrome ===")
print(f'Input: "A man, a plan, a canal: Panama"')
print(f"Result: {is_palindrome('A man, a plan, a canal: Panama')}\n")

print(f'Input: "race a car"')
print(f"Result: {is_palindrome('race a car')}")
````

````python
# Problem 1c: Container With Most Water (LC 11)
# Given heights, find two lines that form the container holding the most water.
#
# Key insight: always move the pointer pointing to the shorter line.
# Moving the taller line can never increase area (width shrinks, height capped by shorter).

def max_area(heights):
    left, right = 0, len(heights) - 1
    best = 0

    while left < right:
        width = right - left
        height = min(heights[left], heights[right])
        area = width * height
        best = max(best, area)

        print(f"  L={left}(h={heights[left]}) R={right}(h={heights[right]}) | width={width} height={height} area={area} | best={best}")

        if heights[left] < heights[right]:
            left += 1
        else:
            right -= 1

    return best


print("=== Container With Most Water ===")
heights = [1, 8, 6, 2, 5, 4, 8, 3, 7]
print(f"Heights: {heights}")
print(f"\nMax area: {max_area(heights)}")
````

## Pattern 2: Sliding Window (Fixed Size)

**When to use**: Find something about ALL contiguous subarrays of size k (max sum, count distinct, etc.)

**Idea**: Build the first window, then *slide* — remove the leftmost element, add the next right element.

```
Array: [ 2, 1, 5, 1, 3, 2 ]    k = 3

Window 1:  [2, 1, 5]  1  3  2     sum = 8
Window 2:   2 [1, 5, 1] 3  2     sum = 8 - 2 + 1 = 7
Window 3:   2  1 [5, 1, 3] 2     sum = 7 - 1 + 3 = 8
Window 4:   2  1  5 [1, 3, 2]    sum = 8 - 5 + 2 = 6
```

Each slide: subtract `nums[i - k]`, add `nums[i]`  

**Time**: O(n)  
**Space**: O(1) for sum, O(k) if tracking elements

````python
# Problem 2a: Maximum Sum Subarray of Size k

def max_sum_subarray(nums, k):
    window_sum = sum(nums[:k])
    best = window_sum
    print(f"  Initial window {nums[:k]} → sum = {window_sum}")

    for i in range(k, len(nums)):
        removed = nums[i - k]
        added = nums[i]
        window_sum = window_sum - removed + added
        best = max(best, window_sum)
        print(f"  Slide: remove {removed}, add {added} → window {nums[i-k+1:i+1]} sum = {window_sum} | best = {best}")

    return best


print("=== Max Sum Subarray of Size k ===")
nums = [2, 1, 5, 1, 3, 2]
k = 3
print(f"Input: {nums}, k={k}")
print(f"Max sum: {max_sum_subarray(nums, k)}")
````

````python
# Problem 2b: Count Distinct Elements in Every Window of Size k

def count_distinct_in_windows(nums, k):
    from collections import defaultdict

    freq = defaultdict(int)
    result = []

    for i in range(k):
        freq[nums[i]] += 1

    result.append(len(freq))
    print(f"  Window {nums[:k]} → distinct = {len(freq)}")

    for i in range(k, len(nums)):
        # Add new element
        freq[nums[i]] += 1

        # Remove old element
        old = nums[i - k]
        freq[old] -= 1
        if freq[old] == 0:
            del freq[old]

        result.append(len(freq))
        print(f"  Window {nums[i-k+1:i+1]} → distinct = {len(freq)}")

    return result


print("=== Count Distinct in Window ===")
nums = [1, 2, 1, 3, 4, 2, 3]
k = 4
print(f"Input: {nums}, k={k}")
print(f"Result: {count_distinct_in_windows(nums, k)}")
````

## Pattern 3: Sliding Window (Variable Size)

**When to use**: Find the longest/shortest subarray satisfying some condition (no repeats, sum ≥ target, at most k distinct chars).

**Idea**: Expand the right pointer to grow the window. When the condition breaks, shrink from the left until valid again.

```
Template:

    left = 0
    for right in range(n):
        # expand: add nums[right] to window state
        while window_is_invalid:
            # shrink: remove nums[left] from window state
            left += 1
        # update answer (window [left..right] is valid)
```

```
Example: longest substring without repeating chars
s = "abcabcbb"

         a  b  c  a  b  c  b  b
Step 1: [a]                         len=1
Step 2: [a  b]                      len=2
Step 3: [a  b  c]                   len=3
Step 4:  a [b  c  a]               'a' repeated → shrink left past old 'a', len=3
Step 5:  a  b [c  a  b]            'b' repeated → shrink, len=3
...
```

**Time**: O(n) — each element enters/leaves the window at most once  
**Space**: O(min(n, alphabet_size))

````python
# Problem 3a: Longest Substring Without Repeating Characters (LC 3)

def length_of_longest_substring(s):
    char_index = {}  # last seen index of each char
    left = 0
    best = 0

    for right in range(len(s)):
        char = s[right]

        if char in char_index and char_index[char] >= left:
            old_left = left
            left = char_index[char] + 1
            print(f"  right={right} char='{char}' REPEAT → shrink left {old_left}→{left} | ", end="")
        else:
            print(f"  right={right} char='{char}' new      | ", end="")

        char_index[char] = right
        window_len = right - left + 1
        best = max(best, window_len)
        print(f"window='{s[left:right+1]}' len={window_len} best={best}")

    return best


print("=== Longest Substring Without Repeating Characters ===")
s = "abcabcbb"
print(f'Input: "{s}"')
print(f"Result: {length_of_longest_substring(s)}\n")

s = "pwwkew"
print(f'Input: "{s}"')
print(f"Result: {length_of_longest_substring(s)}")
````

````python
# Problem 3b: Minimum Size Subarray Sum (LC 209)
# Find shortest subarray with sum >= target

def min_subarray_len(target, nums):
    left = 0
    current_sum = 0
    best = float('inf')

    for right in range(len(nums)):
        current_sum += nums[right]
        print(f"  Expand right={right} add {nums[right]} → sum={current_sum}")

        while current_sum >= target:
            window_len = right - left + 1
            best = min(best, window_len)
            print(f"    sum≥{target} → window[{left}:{right+1}]={nums[left:right+1]} len={window_len} best={best} | shrink left")
            current_sum -= nums[left]
            left += 1

    return best if best != float('inf') else 0


print("=== Minimum Size Subarray Sum ===")
nums = [2, 3, 1, 2, 4, 3]
target = 7
print(f"Input: {nums}, target={target}")
print(f"Min length: {min_subarray_len(target, nums)}")
````

## Pattern 4: Prefix Sum

**When to use**: You need to answer many range-sum queries, or find subarrays whose sum equals a target.

**Idea**: Build a prefix sum array where `prefix[i] = sum(nums[0..i-1])`. Then:
- Sum of `nums[i..j]` = `prefix[j+1] - prefix[i]` in O(1)

```
Index:      0   1   2   3   4   5
nums:     [ 3,  1,  4,  1,  5,  9 ]

prefix:  [0,  3,  4,  8,  9, 14, 23]
          ^                          ^
       prefix[0]=0            prefix[6]=sum of all

Sum of nums[2..4] = prefix[5] - prefix[2] = 14 - 4 = 10
Check: nums[2]+nums[3]+nums[4] = 4+1+5 = 10  ✓
```

**For "subarray sum = k"**: If `prefix[j] - prefix[i] = k`, there's a subarray `[i..j-1]` with sum k.  
Use a hash map to count how many times each prefix sum has occurred → O(n) solution.

**Time**: O(n) build, O(1) per query  
**Space**: O(n)

````python
# Problem 4a: Range Sum Query — Immutable (LC 303)

class NumArray:
    def __init__(self, nums):
        self.prefix = [0] * (len(nums) + 1)
        for i in range(len(nums)):
            self.prefix[i + 1] = self.prefix[i] + nums[i]

    def sum_range(self, left, right):
        return self.prefix[right + 1] - self.prefix[left]


print("=== Range Sum Query ===")
nums = [-2, 0, 3, -5, 2, -1]
print(f"nums: {nums}")
na = NumArray(nums)
print(f"prefix: {na.prefix}\n")

queries = [(0, 2), (2, 5), (0, 5)]
for l, r in queries:
    result = na.sum_range(l, r)
    print(f"  sum_range({l}, {r}) = prefix[{r+1}] - prefix[{l}] = {na.prefix[r+1]} - {na.prefix[l]} = {result}")
````

````python
# Problem 4b: Subarray Sum Equals K (LC 560)
# Count subarrays whose sum = k
#
# Trick: if prefix[j] - k appeared before as some prefix[i],
# then subarray [i..j-1] sums to k.

def subarray_sum(nums, k):
    from collections import defaultdict

    prefix_count = defaultdict(int)
    prefix_count[0] = 1  # empty prefix
    current_sum = 0
    count = 0

    for i, num in enumerate(nums):
        current_sum += num
        need = current_sum - k

        found = prefix_count[need]
        count += found
        print(f"  i={i} num={num} | prefix_sum={current_sum} | need={need} | found {found} times → count={count}")

        prefix_count[current_sum] += 1

    return count


print("=== Subarray Sum Equals K ===")
nums = [1, 1, 1]
k = 2
print(f"Input: {nums}, k={k}")
print(f"Count: {subarray_sum(nums, k)}\n")

nums = [1, 2, 3, -2, 5]
k = 3
print(f"Input: {nums}, k={k}")
print(f"Count: {subarray_sum(nums, k)}")
````

## Pattern 5: Kadane's Algorithm

**When to use**: Find the contiguous subarray with the maximum (or minimum) sum.

**Idea**: At each position, make a decision: "Should I extend the current streak, or start fresh here?"

```
current_max = max(nums[i], current_max + nums[i])

If current_max + nums[i] < nums[i]  →  previous streak was negative, start fresh
Otherwise                            →  extend the streak
```

```
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

i=0: num=-2  current=max(-2, -2)=-2         global=-2  (start fresh)
i=1: num= 1  current=max(1, -2+1=-1)= 1    global= 1  (start fresh)
i=2: num=-3  current=max(-3, 1-3=-2)=-2     global= 1  (extend)
i=3: num= 4  current=max(4, -2+4=2)= 4     global= 4  (start fresh)
i=4: num=-1  current=max(-1, 4-1=3)= 3     global= 4  (extend)
i=5: num= 2  current=max(2, 3+2=5)= 5      global= 5  (extend)
i=6: num= 1  current=max(1, 5+1=6)= 6      global= 6  (extend) ← answer
i=7: num=-5  current=max(-5, 6-5=1)= 1     global= 6  (extend)
i=8: num= 4  current=max(4, 1+4=5)= 5      global= 6  (extend)

Answer: 6  (subarray [4, -1, 2, 1])
```

**Time**: O(n)  
**Space**: O(1)

````python
# Problem 5: Maximum Subarray (LC 53)

def max_subarray(nums):
    current_max = nums[0]
    global_max = nums[0]

    print(f"  i=0  num={nums[0]:3d}  current={current_max:3d}  global={global_max:3d}  (start)")

    for i in range(1, len(nums)):
        extend = current_max + nums[i]
        fresh = nums[i]
        decision = "extend" if extend >= fresh else "START FRESH"

        current_max = max(fresh, extend)
        global_max = max(global_max, current_max)

        print(f"  i={i}  num={nums[i]:3d}  current={current_max:3d}  global={global_max:3d}  ({decision})")

    return global_max


print("=== Kadane's Algorithm — Maximum Subarray ===")
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
print(f"Input: {nums}")
print(f"\nMax subarray sum: {max_subarray(nums)}")
````

## Pattern 6: Dutch National Flag

**When to use**: Partition an array into exactly 3 groups in a single pass (e.g., sort 0s/1s/2s).

**Idea**: Maintain 3 pointers:
- `low`: boundary of 0s (everything before low is 0)
- `mid`: current element being examined
- `high`: boundary of 2s (everything after high is 2)

```
Regions:

  [ 0 0 0 | 1 1 1 | ? ? ? | 2 2 2 ]
    ^^^^^^   ^^^^^^   ^^^^^   ^^^^^^
   all 0s   all 1s   unseen  all 2s
           ^         ^     ^
          low       mid   high

Rules:
  nums[mid] == 0  →  swap(low, mid), low++, mid++
  nums[mid] == 1  →  mid++  (already in right spot)
  nums[mid] == 2  →  swap(mid, high), high--  (don't advance mid — swapped value unknown)
```

**Time**: O(n)  
**Space**: O(1)

````python
# Problem 6: Sort Colors (LC 75)

def sort_colors(nums):
    low, mid, high = 0, 0, len(nums) - 1

    while mid <= high:
        print(f"  low={low} mid={mid} high={high} | nums={nums}", end="")

        if nums[mid] == 0:
            nums[low], nums[mid] = nums[mid], nums[low]
            print(f"  → swap(low,mid) → {nums}")
            low += 1
            mid += 1
        elif nums[mid] == 1:
            print(f"  → mid++ (1 in place)")
            mid += 1
        else:
            nums[mid], nums[high] = nums[high], nums[mid]
            print(f"  → swap(mid,high) → {nums}")
            high -= 1

    return nums


print("=== Sort Colors (Dutch National Flag) ===")
nums = [2, 0, 2, 1, 1, 0]
print(f"Input:  {nums}")
print(f"Output: {sort_colors(nums.copy())}")
````

## Pattern 7: Merge Intervals

**When to use**: Given a list of intervals, merge all overlapping ones.

**Idea**:
1. Sort intervals by start time
2. Iterate: if the current interval overlaps with the last merged one, extend it. Otherwise, start a new merged interval.

```
Sorted:  [1,3]  [2,6]  [8,10]  [15,18]

merged = [[1,3]]
  [2,6]: 2 <= 3 (overlaps) → extend to [1,6]
  [8,10]: 8 > 6 (no overlap) → add [8,10]
  [15,18]: 15 > 10 → add [15,18]

Result: [[1,6], [8,10], [15,18]]
```

**Time**: O(n log n) for sort  
**Space**: O(n) for output

````python
# Problem 7: Merge Intervals (LC 56)

def merge_intervals(intervals):
    intervals.sort(key=lambda x: x[0])
    merged = [intervals[0]]
    print(f"  Sorted: {intervals}")
    print(f"  Start with: {merged}")

    for current in intervals[1:]:
        last = merged[-1]

        if current[0] <= last[1]:
            merged[-1] = [last[0], max(last[1], current[1])]
            print(f"  {current} overlaps with {last} → merge to {merged[-1]}")
        else:
            merged.append(current)
            print(f"  {current} no overlap → add new")

    return merged


print("=== Merge Intervals ===")
intervals = [[1, 3], [2, 6], [8, 10], [15, 18]]
print(f"Input: {intervals}")
print(f"Merged: {merge_intervals(intervals)}\n")

intervals = [[1, 4], [4, 5]]
print(f"Input: {intervals}")
print(f"Merged: {merge_intervals(intervals)}")
````

## Pattern 8: Anagram Checking & Grouping

**When to use**: Check if two strings are anagrams, or group strings by their character composition.

**Idea**: Two strings are anagrams if they have the same character frequency.

```
"listen"  →  Counter: {l:1, i:1, s:1, t:1, e:1, n:1}
"silent"  →  Counter: {s:1, i:1, l:1, e:1, n:1, t:1}
Same! → anagram ✓

For grouping: use sorted string as key
"eat" → "aet"    "tea" → "aet"    "ate" → "aet"    → all in same group
"tan" → "ant"    "nat" → "ant"                      → same group
```

````python
# Problem 8a: Valid Anagram (LC 242)

from collections import Counter

def is_anagram(s, t):
    count_s = Counter(s)
    count_t = Counter(t)
    print(f"  '{s}' → {dict(count_s)}")
    print(f"  '{t}' → {dict(count_t)}")
    result = count_s == count_t
    print(f"  Same? {result}")
    return result


print("=== Valid Anagram ===")
is_anagram("anagram", "nagaram")
print()
is_anagram("rat", "car")
````

````python
# Problem 8b: Group Anagrams (LC 49)

from collections import defaultdict

def group_anagrams(strs):
    groups = defaultdict(list)

    for s in strs:
        key = ''.join(sorted(s))
        groups[key].append(s)
        print(f"  '{s}' → key='{key}'")

    print(f"\n  Groups:")
    for key, words in groups.items():
        print(f"    '{key}' → {words}")

    return list(groups.values())


print("=== Group Anagrams ===")
strs = ["eat", "tea", "tan", "ate", "nat", "bat"]
print(f"Input: {strs}\n")
result = group_anagrams(strs)
print(f"\nResult: {result}")
````

## Pattern 9: Stock Buy & Sell

**When to use**: Given daily prices, maximize profit from buying and selling.

**Problem 9a — One Transaction**  
Track the minimum price seen so far. At each day, compute profit if you sold today. Keep the maximum.

```
prices = [7, 1, 5, 3, 6, 4]

Day 0: price=7  min_price=7  profit=0   max_profit=0
Day 1: price=1  min_price=1  profit=0   max_profit=0
Day 2: price=5  min_price=1  profit=4   max_profit=4  ← buy@1, sell@5
Day 3: price=3  min_price=1  profit=2   max_profit=4
Day 4: price=6  min_price=1  profit=5   max_profit=5  ← buy@1, sell@6 ✓
Day 5: price=4  min_price=1  profit=3   max_profit=5
```

**Problem 9b — Unlimited Transactions**  
Collect every uphill: if `prices[i] > prices[i-1]`, add the difference.

````python
# Problem 9a: Best Time to Buy and Sell Stock (LC 121) — one transaction

def max_profit_one(prices):
    min_price = prices[0]
    max_profit = 0

    for i, price in enumerate(prices):
        min_price = min(min_price, price)
        profit = price - min_price
        max_profit = max(max_profit, profit)
        print(f"  Day {i}: price={price}  min_so_far={min_price}  profit_if_sell={profit}  max_profit={max_profit}")

    return max_profit


print("=== Stock Buy & Sell — One Transaction ===")
prices = [7, 1, 5, 3, 6, 4]
print(f"Prices: {prices}")
print(f"Max profit: {max_profit_one(prices)}")
````

````python
# Problem 9b: Best Time to Buy and Sell Stock II (LC 122) — unlimited transactions

def max_profit_unlimited(prices):
    total = 0

    for i in range(1, len(prices)):
        diff = prices[i] - prices[i - 1]
        if diff > 0:
            total += diff
            print(f"  Day {i}: {prices[i-1]}→{prices[i]}  gain={diff}  total={total}")
        else:
            print(f"  Day {i}: {prices[i-1]}→{prices[i]}  skip (downhill)")

    return total


print("=== Stock Buy & Sell — Unlimited Transactions ===")
prices = [7, 1, 5, 3, 6, 4]
print(f"Prices: {prices}")
print(f"Max profit: {max_profit_unlimited(prices)}")
````

## Bonus: Rotate Array

**When to use**: Rotate an array by k positions to the right, in-place.

**The Reversal Trick**: Three reverses accomplish a rotation without extra space.

```
nums = [1, 2, 3, 4, 5, 6, 7]   k = 3

Step 1: Reverse entire array    → [7, 6, 5, 4, 3, 2, 1]
Step 2: Reverse first k=3       → [5, 6, 7, 4, 3, 2, 1]
Step 3: Reverse remaining n-k=4 → [5, 6, 7, 1, 2, 3, 4]  ✓
```

**Why it works**: Reversing "undoes" the ordering within each half, and reversing the whole array first puts the halves in the right position.

**Time**: O(n)  
**Space**: O(1)

````python
# Bonus: Rotate Array (LC 189) — Reversal Trick

def reverse_section(nums, start, end):
    while start < end:
        nums[start], nums[end] = nums[end], nums[start]
        start += 1
        end -= 1


def rotate_array(nums, k):
    n = len(nums)
    k = k % n  # handle k > n

    print(f"  Original:               {nums}")

    reverse_section(nums, 0, n - 1)
    print(f"  After reverse all:      {nums}")

    reverse_section(nums, 0, k - 1)
    print(f"  After reverse [0:{k}]:    {nums}")

    reverse_section(nums, k, n - 1)
    print(f"  After reverse [{k}:{n}]:    {nums}")

    return nums


print("=== Rotate Array ===")
nums = [1, 2, 3, 4, 5, 6, 7]
k = 3
print(f"Input: {nums}, k={k}\n")
print(f"\nResult: {rotate_array(nums, k)}")
````
