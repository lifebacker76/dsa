## Pattern 1: Two Pointers

**When to use**: Sorted array, finding pairs, palindrome checks, comparing from both ends.

**Idea**: Place one pointer at the start, one at the end. Move them toward each other based on a condition.

```
Array: [1, 3, 5, 7, 9, 11]
        ^                ^  
       left            right
```

**Time**: O(n) | **Space**: O(1)

````python
# Problem 1a: Two Sum II (sorted array)
# Given a SORTED array and a target, find two numbers that add up to target.
# Return their 1-based indices.

def two_sum_sorted(nums, target):
    left, right = 0, len(nums) - 1
    
    while left < right:
        current_sum = nums[left] + nums[right]
        
        if current_sum == target:
            return [left + 1, right + 1]
        elif current_sum < target:
            left += 1       # need a bigger sum → move left pointer right
        else:
            right -= 1      # need a smaller sum → move right pointer left
    
    return []

# Trace through it
nums = [2, 7, 11, 15]
target = 9
print(f"Input: {nums}, target={target}")
print(f"Step 1: left=0(2), right=3(15), sum=17 > 9 → move right")
print(f"Step 2: left=0(2), right=2(11), sum=13 > 9 → move right")
print(f"Step 3: left=0(2), right=1(7),  sum=9  = 9 → FOUND!")
print(f"Result: {two_sum_sorted(nums, target)}")
````

````python
# Problem 1b: Valid Palindrome
# Check if a string is a palindrome, considering only alphanumeric characters.

def is_palindrome(s):
    left, right = 0, len(s) - 1
    
    while left < right:
        # Skip non-alphanumeric characters
        while left < right and not s[left].isalnum():
            left += 1
        while left < right and not s[right].isalnum():
            right -= 1
        
        if s[left].lower() != s[right].lower():
            return False
        
        left += 1
        right -= 1
    
    return True

print(is_palindrome("A man, a plan, a canal: Panama"))  # True
print(is_palindrome("race a car"))                       # False
print(is_palindrome("Was it a car or a cat I saw?"))     # True
````

````python
# Problem 1c: Container With Most Water
# Given heights, find two lines that form a container holding the most water.
#
#   |         |              
#   |         |     |        
#   |   |     |     |        
#   |   |  |  |     |        
#   |   |  |  |  |  |        
#   1   8  6  2  5  4  8  3  7

def max_area(heights):
    left, right = 0, len(heights) - 1
    max_water = 0
    
    while left < right:
        width = right - left
        height = min(heights[left], heights[right])
        water = width * height
        max_water = max(max_water, water)
        
        # Move the shorter line inward (greedy choice)
        if heights[left] < heights[right]:
            left += 1
        else:
            right -= 1
    
    return max_water

print(max_area([1, 8, 6, 2, 5, 4, 8, 3, 7]))  # 49
````

## Pattern 2: Sliding Window (Fixed Size)

**When to use**: Find something in all contiguous subarrays of size `k`.

**Idea**: Instead of recalculating from scratch each time, slide the window by removing the leftmost element and adding the next element.

```
Array: [2, 1, 5, 1, 3, 2], k=3

Window 1: [2, 1, 5] → sum = 8
Window 2: [1, 5, 1] → sum = 8 - 2 + 1 = 7  (slide!)
Window 3: [5, 1, 3] → sum = 7 - 1 + 3 = 9  (slide!)
Window 4: [1, 3, 2] → sum = 9 - 5 + 2 = 6  (slide!)
```

**Time**: O(n) | **Space**: O(1)

````python
# Problem 2a: Maximum sum of subarray of size k

def max_sum_subarray(nums, k):
    # Calculate sum of first window
    window_sum = sum(nums[:k])
    max_sum = window_sum
    
    # Slide the window
    for i in range(k, len(nums)):
        window_sum += nums[i] - nums[i - k]  # add new, remove old
        max_sum = max(max_sum, window_sum)
    
    return max_sum

nums = [2, 1, 5, 1, 3, 2]
k = 3
print(f"Max sum of subarray of size {k}: {max_sum_subarray(nums, k)}")  # 9

# Let's trace it
print("\nTracing:")
window_sum = sum(nums[:k])
print(f"  Window [0:3] = {nums[0:3]} → sum = {window_sum}")
for i in range(k, len(nums)):
    removed = nums[i - k]
    added = nums[i]
    window_sum += added - removed
    print(f"  Window [{i-k+1}:{i+1}] = {nums[i-k+1:i+1]} → removed {removed}, added {added} → sum = {window_sum}")
````

````python
# Problem 2b: Count distinct elements in every window of size k
# (This is on your topics list!)

from collections import defaultdict

def count_distinct_in_windows(nums, k):
    freq = defaultdict(int)
    result = []
    
    for i in range(len(nums)):
        freq[nums[i]] += 1
        
        # Remove element going out of window
        if i >= k:
            left_elem = nums[i - k]
            freq[left_elem] -= 1
            if freq[left_elem] == 0:
                del freq[left_elem]
        
        # Window is complete
        if i >= k - 1:
            result.append(len(freq))
    
    return result

nums = [1, 2, 1, 3, 4, 2, 3]
k = 4
print(f"Distinct counts in windows of {k}: {count_distinct_in_windows(nums, k)}")
# Windows: [1,2,1,3]→3, [2,1,3,4]→4, [1,3,4,2]→4, [3,4,2,3]→3
````

## Pattern 3: Sliding Window (Variable Size)

**When to use**: Find the longest/shortest subarray/substring satisfying a condition.

**Idea**: Expand the right pointer to grow the window. When the condition breaks, shrink from the left.

```
Template:
  left = 0
  for right in range(n):
      # expand: add nums[right] to window
      while window_is_invalid:
          # shrink: remove nums[left] from window
          left += 1
      # update answer
```

````python
# Problem 3a: Longest Substring Without Repeating Characters (LeetCode #3)
# One of the MOST asked interview questions!

def length_of_longest_substring(s):
    char_set = set()
    left = 0
    max_len = 0
    
    for right in range(len(s)):
        # If char already in window, shrink from left
        while s[right] in char_set:
            char_set.remove(s[left])
            left += 1
        
        char_set.add(s[right])
        max_len = max(max_len, right - left + 1)
    
    return max_len

# Let's trace it step by step
s = "abcabcbb"
print(f"Input: '{s}'")
print(f"Result: {length_of_longest_substring(s)}")  # 3 ("abc")
print()

# Detailed trace
char_set = set()
left = 0
for right in range(len(s)):
    while s[right] in char_set:
        print(f"  Shrink: remove '{s[left]}', left → {left+1}")
        char_set.remove(s[left])
        left += 1
    char_set.add(s[right])
    window = s[left:right+1]
    print(f"  right={right}, char='{s[right]}', window='{window}', len={right-left+1}")
````

````python
# Problem 3b: Minimum Size Subarray Sum
# Find the smallest subarray with sum >= target

def min_subarray_len(target, nums):
    left = 0
    window_sum = 0
    min_len = float('inf')
    
    for right in range(len(nums)):
        window_sum += nums[right]
        
        while window_sum >= target:
            min_len = min(min_len, right - left + 1)
            window_sum -= nums[left]
            left += 1
    
    return min_len if min_len != float('inf') else 0

print(min_subarray_len(7, [2, 3, 1, 2, 4, 3]))  # 2 → [4, 3]
print(min_subarray_len(11, [1, 1, 1, 1, 1]))     # 0 → impossible
````

## Pattern 4: Prefix Sum

**When to use**: Range sum queries, subarray sum problems.  
**Idea**: Pre-compute cumulative sums so any range sum is a single subtraction.

```
Array:      [3, 1, 4, 1, 5]
Prefix Sum: [0, 3, 4, 8, 9, 14]

Sum(1..3) = prefix[4] - prefix[1] = 9 - 3 = 6  ← O(1)!
```

**Analogy**: Like a running total column in Excel / `cumsum()` in pandas.

````python
# Problem 4a: Range Sum Query

def build_prefix_sum(nums):
    prefix = [0] * (len(nums) + 1)
    for i in range(len(nums)):
        prefix[i + 1] = prefix[i] + nums[i]
    return prefix

def range_sum(prefix, left, right):
    """Sum of nums[left..right] inclusive, in O(1)."""
    return prefix[right + 1] - prefix[left]

nums = [3, 1, 4, 1, 5, 9, 2, 6]
prefix = build_prefix_sum(nums)

print(f"Array:      {nums}")
print(f"Prefix Sum: {prefix}")
print(f"Sum(1..3):  {range_sum(prefix, 1, 3)}")  # 1+4+1 = 6
print(f"Sum(0..4):  {range_sum(prefix, 0, 4)}")  # 3+1+4+1+5 = 14
print(f"Sum(5..7):  {range_sum(prefix, 5, 7)}")  # 9+2+6 = 17
````

````python
# Problem 4b: Subarray Sum Equals K (LeetCode #560)
# Count how many subarrays sum to exactly k.
# This combines prefix sum with a hash map!

def subarray_sum(nums, k):
    count = 0
    current_sum = 0
    # prefix_sum_count stores: how many times each prefix sum has occurred
    prefix_count = {0: 1}  # empty prefix has sum 0
    
    for num in nums:
        current_sum += num
        
        # If (current_sum - k) was a previous prefix sum,
        # then the subarray between them sums to k
        if current_sum - k in prefix_count:
            count += prefix_count[current_sum - k]
        
        prefix_count[current_sum] = prefix_count.get(current_sum, 0) + 1
    
    return count

print(subarray_sum([1, 1, 1], 2))        # 2 → [1,1] and [1,1]
print(subarray_sum([1, 2, 3], 3))        # 2 → [1,2] and [3]
print(subarray_sum([1, -1, 1, 1, -1], 1))  # 5
````

## Pattern 5: Kadane's Algorithm

**Problem**: Find the contiguous subarray with the maximum sum.  
**When to use**: Any "maximum subarray" variant.

**Idea**: At each position, decide: extend the current subarray, or start fresh here?

```
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

At each step: current_max = max(nums[i], current_max + nums[i])
  "Should I continue the streak or start a new one?"
```

````python
def max_subarray(nums):
    current_max = nums[0]
    global_max = nums[0]
    
    for i in range(1, len(nums)):
        # Either extend the previous subarray or start fresh
        current_max = max(nums[i], current_max + nums[i])
        global_max = max(global_max, current_max)
    
    return global_max

# Trace through it
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
print(f"Input: {nums}")
print(f"Max subarray sum: {max_subarray(nums)}")  # 6 → [4, -1, 2, 1]

print("\nStep-by-step trace:")
current_max = nums[0]
global_max = nums[0]
print(f"  i=0: num={nums[0]:>3}, current_max={current_max:>3}, global_max={global_max:>3}")
for i in range(1, len(nums)):
    extend = current_max + nums[i]
    fresh = nums[i]
    choice = "extend" if extend >= fresh else "start fresh"
    current_max = max(fresh, extend)
    global_max = max(global_max, current_max)
    print(f"  i={i}: num={nums[i]:>3}, extend={extend:>3}, fresh={fresh:>3} → {choice:>11}, current_max={current_max:>3}, global_max={global_max:>3}")
````

## Pattern 6: Dutch National Flag (Sort 0, 1, 2)

**Problem**: Sort an array containing only 0s, 1s, and 2s — in one pass, O(1) space.  
**When to use**: 3-way partitioning, asked frequently in interviews.

**Idea**: Use 3 pointers — `low`, `mid`, `high`.
- Everything before `low` = 0
- Everything between `low` and `mid` = 1  
- Everything after `high` = 2
- `mid` scans through the unknown region

````python
def sort_colors(nums):
    low, mid, high = 0, 0, len(nums) - 1
    
    while mid <= high:
        if nums[mid] == 0:
            nums[low], nums[mid] = nums[mid], nums[low]
            low += 1
            mid += 1
        elif nums[mid] == 1:
            mid += 1
        else:  # nums[mid] == 2
            nums[mid], nums[high] = nums[high], nums[mid]
            high -= 1
            # Don't increment mid! We need to check the swapped element
    
    return nums

# Test
test = [2, 0, 2, 1, 1, 0]
print(f"Before: {test}")
sort_colors(test)
print(f"After:  {test}")  # [0, 0, 1, 1, 2, 2]

# Trace
print("\nTracing [2, 0, 1, 2, 0, 1]:")
nums = [2, 0, 1, 2, 0, 1]
low, mid, high = 0, 0, len(nums) - 1
step = 0
while mid <= high:
    step += 1
    if nums[mid] == 0:
        nums[low], nums[mid] = nums[mid], nums[low]
        print(f"  Step {step}: nums[mid]=0, swap low↔mid → {nums}, low={low+1}, mid={mid+1}")
        low += 1; mid += 1
    elif nums[mid] == 1:
        print(f"  Step {step}: nums[mid]=1, skip        → {nums}, mid={mid+1}")
        mid += 1
    else:
        nums[mid], nums[high] = nums[high], nums[mid]
        print(f"  Step {step}: nums[mid]=2, swap mid↔high → {nums}, high={high-1}")
        high -= 1
````

## Pattern 7: Merge Intervals

**Problem**: Given a list of intervals, merge all overlapping ones.  
**When to use**: Scheduling, calendar problems, range merging.

```
Input:  [[1,3], [2,6], [8,10], [15,18]]
Output: [[1,6], [8,10], [15,18]]
        (1,3) and (2,6) overlap → merged to (1,6)
```

````python
def merge_intervals(intervals):
    intervals.sort(key=lambda x: x[0])  # Sort by start time
    merged = [intervals[0]]
    
    for current in intervals[1:]:
        last = merged[-1]
        
        if current[0] <= last[1]:  # Overlapping
            last[1] = max(last[1], current[1])  # Extend end
        else:
            merged.append(current)  # No overlap, add new
    
    return merged

intervals = [[1,3], [2,6], [8,10], [15,18]]
print(f"Input:  {intervals}")
print(f"Merged: {merge_intervals(intervals)}")

intervals2 = [[1,4], [0,4]]
print(f"\nInput:  {intervals2}")
print(f"Merged: {merge_intervals(intervals2)}")
````

## Pattern 8: Anagram Checking & Grouping

**Anagram**: Two strings with the same characters in different order.  
Example: "listen" and "silent"

````python
from collections import Counter, defaultdict

# Problem 8a: Check if two strings are anagrams
def is_anagram(s, t):
    return Counter(s) == Counter(t)

print(is_anagram("listen", "silent"))  # True
print(is_anagram("hello", "world"))    # False

# Problem 8b: Group Anagrams (LeetCode #49)
def group_anagrams(strs):
    groups = defaultdict(list)
    for s in strs:
        key = tuple(sorted(s))  # sorted chars as key
        groups[key].append(s)
    return list(groups.values())

words = ["eat", "tea", "tan", "ate", "nat", "bat"]
print(f"\nGrouped anagrams:")
for group in group_anagrams(words):
    print(f"  {group}")
````

## Pattern 9: Stock Buy & Sell

**Problem**: Given daily stock prices, find the best time to buy and sell for max profit.

````python
# Problem 9a: Best Time to Buy and Sell Stock (one transaction)
def max_profit(prices):
    min_price = float('inf')
    max_profit = 0
    
    for price in prices:
        min_price = min(min_price, price)         # track lowest price so far
        max_profit = max(max_profit, price - min_price)  # profit if sold today
    
    return max_profit

prices = [7, 1, 5, 3, 6, 4]
print(f"Prices: {prices}")
print(f"Max profit: {max_profit(prices)}")  # 5 (buy at 1, sell at 6)

# Trace
min_p = float('inf')
for i, p in enumerate(prices):
    min_p = min(min_p, p)
    profit = p - min_p
    print(f"  Day {i}: price={p}, min_so_far={min_p}, profit_if_sell={profit}")
````

````python
# Problem 9b: Best Time to Buy and Sell Stock II (unlimited transactions)
def max_profit_ii(prices):
    profit = 0
    for i in range(1, len(prices)):
        if prices[i] > prices[i-1]:
            profit += prices[i] - prices[i-1]  # capture every upward move
    return profit

print(f"Unlimited trades: {max_profit_ii([7, 1, 5, 3, 6, 4])}")  # 7
````

## Bonus: Rotate Array

**Problem**: Rotate an array to the right by k steps.

````python
def rotate_array(nums, k):
    """Rotate using the reversal trick — O(n) time, O(1) space."""
    n = len(nums)
    k = k % n  # handle k > n
    
    def reverse(left, right):
        while left < right:
            nums[left], nums[right] = nums[right], nums[left]
            left += 1
            right -= 1
    
    # Step 1: Reverse entire array
    # Step 2: Reverse first k elements  
    # Step 3: Reverse remaining elements
    reverse(0, n - 1)
    reverse(0, k - 1)
    reverse(k, n - 1)
    return nums

nums = [1, 2, 3, 4, 5, 6, 7]
k = 3
print(f"Original: {nums}")
print(f"After rotating right by {k}:")

# Show each step
demo = [1, 2, 3, 4, 5, 6, 7]
demo.reverse()
print(f"  1. Reverse all:      {demo}")
demo[:k] = reversed(demo[:k])
print(f"  2. Reverse first {k}:  {demo}")
demo[k:] = reversed(demo[k:])
print(f"  3. Reverse rest:     {demo}")

print(f"\nResult: {rotate_array([1,2,3,4,5,6,7], 3)}")
````
