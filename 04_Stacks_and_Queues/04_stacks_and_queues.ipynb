## Why Stacks & Queues?

They're everywhere in computing:
- **Stack**: Undo button (Ctrl+Z), browser back button, function call stack, expression evaluation
- **Queue**: Print queue, message processing, breadth-first search

| | Stack | Queue |
|---|---|---|
| Order | LIFO (Last In, First Out) | FIFO (First In, First Out) |
| Analogy | Stack of plates | Line at a checkout counter |
| Add | O(1) — push to top | O(1) — add to back |
| Remove | O(1) — pop from top | O(1) — remove from front |
| Peek | O(1) — look at top | O(1) — look at front |

```
STACK (pile of plates):        QUEUE (checkout line):

  ┌─────┐                      
  │  30 │ ← top (push/pop)    FRONT → [10] [20] [30] ← BACK
  ├─────┤                      remove ←              ← add
  │  20 │                     
  ├─────┤                     
  │  10 │ ← bottom            
  └─────┘                     
```

Two data structures, two philosophies:
- **Stack**: "Deal with the most RECENT thing first" — like a stack of plates, you always grab the top one.
- **Queue**: "Deal with things in the ORDER they arrived" — like a line at a store, first come first served.

## Part 1: THE STACK

### The Mental Model

Imagine a stack of plates in a cafeteria:

```
  You can ONLY touch the TOP plate.

  ┌─────────┐
  │ plate C │  ← This is the ONLY plate you can grab
  ├─────────┤
  │ plate B │  ← Can't touch this until C is removed
  ├─────────┤
  │ plate A │  ← Buried at the bottom
  └─────────┘

  LIFO = Last In, First Out
  Plate C went on LAST, comes off FIRST.
```

### Three Core Operations

```
  push(x)          pop()              peek()
  Add to top       Remove from top    Look at top

   ┌───┐            ┌───┐
   │ x │ ↓ push     │ C │ ↑ pop       → C  (just look)
  ┌┴───┴┐          ┌─────┐           ┌─────┐
  │  C  │          │  B  │           │  C  │
  ├─────┤          ├─────┤           ├─────┤
  │  B  │          │  A  │           │  B  │
  ├─────┤          └─────┘           ├─────┤
  │  A  │                            │  A  │
  └─────┘                            └─────┘

  All O(1) — we ONLY ever touch the top.
```

**Key rule**: You can ONLY interact with the top element. No peeking at the middle, no removing from the bottom. That constraint is what makes stacks so useful — they naturally handle "nesting" and "undo" patterns.

````python
class Stack:
    def __init__(self):
        self._data = []

    def push(self, val):
        self._data.append(val)

    def pop(self):
        if self.is_empty():
            raise IndexError("pop from empty stack")
        return self._data.pop()

    def peek(self):
        if self.is_empty():
            raise IndexError("peek at empty stack")
        return self._data[-1]

    def is_empty(self):
        return len(self._data) == 0

    def size(self):
        return len(self._data)

    def __repr__(self):
        return f"Stack {self._data}"


s = Stack()
for val in [10, 20, 30]:
    s.push(val)
    print(f"push({val}): {s}")

print(f"peek():   {s.peek()}")

val = s.pop()
print(f"pop():    {val} → {s}")

val = s.pop()
print(f"pop():    {val} → {s}")

print(f"size():   {s.size()}")
print(f"empty?:   {s.is_empty()}")
````

## Part 2: THE QUEUE

### The Mental Model

Imagine a line at a checkout counter:

```
  First person in line gets served first.

  FRONT                              BACK
    ↓                                  ↓
  ┌──────┐  ┌──────┐  ┌──────┐  ┌──────┐
  │ Ann  │──│ Bob  │──│ Carl │──│ Dana │
  └──────┘  └──────┘  └──────┘  └──────┘
    ↑                                ↑
  dequeue                         enqueue
  (serve)                       (join line)

  FIFO = First In, First Out
  Ann arrived FIRST, gets served FIRST.
```

### Three Core Operations

```
  enqueue(x)  →  add to the BACK of the line
  dequeue()   →  remove from the FRONT of the line
  peek()      →  look at who's at the FRONT (without removing)

  All O(1).
```

**Key distinction from stack**: A stack has ONE end (the top). A queue has TWO ends — you add at the back and remove from the front.

````python
from collections import deque


class Queue:
    def __init__(self):
        self._data = deque()

    def enqueue(self, val):
        self._data.append(val)

    def dequeue(self):
        if self.is_empty():
            raise IndexError("dequeue from empty queue")
        return self._data.popleft()

    def peek(self):
        if self.is_empty():
            raise IndexError("peek at empty queue")
        return self._data[0]

    def is_empty(self):
        return len(self._data) == 0

    def size(self):
        return len(self._data)

    def __repr__(self):
        return f"Queue {list(self._data)}"


q = Queue()
for val in [10, 20, 30]:
    q.enqueue(val)
    print(f"enqueue({val}): {q}")

print(f"peek():      {q.peek()}")

val = q.dequeue()
print(f"dequeue():   {val} → {q}")

val = q.dequeue()
print(f"dequeue():   {val} → {q}")

print(f"size():      {q.size()}")
print(f"empty?:      {q.is_empty()}")
````

### Why `deque` Instead of `list`?

You might wonder: why not just use a Python list and call `list.pop(0)` to dequeue?

```
list.pop(0):

  Before:  [10, 20, 30, 40]
             ↑ remove this

  After:   [20, 30, 40, __]
            ←── ←── ←──       ← SHIFT everything left! O(n)


deque.popleft():

  Just update the front pointer → O(1)
  No shifting needed!
```

| Operation | `list` | `deque` |
|---|---|---|
| `append` (add to back) | O(1) | O(1) |
| `pop()` (remove from back) | O(1) | O(1) |
| `pop(0)` / `popleft()` (remove from front) | **O(n)** | **O(1)** |

For a queue, we always remove from the front — so `deque` is the right choice.

## Core Problem 1: Valid Parentheses

**LeetCode 20** — Given a string containing just `(`, `)`, `{`, `}`, `[`, `]`, determine if it's valid.

### Why a Stack?

Brackets NEST — the most recently opened bracket must be closed first. That's exactly LIFO!

**Algorithm**: Scan left to right. Open bracket → push. Close bracket → pop and check if it matches.

### Step-by-Step Trace: `"({[]})"`

```
char '('  → open, push     → stack: ['(']
char '{'  → open, push     → stack: ['(', '{']
char '['  → open, push     → stack: ['(', '{', '[']
char ']'  → close, pop '[' → matches '['? YES → stack: ['(', '{']
char '}'  → close, pop '{' → matches '{'? YES → stack: ['(']
char ')'  → close, pop '(' → matches '('? YES → stack: []

Stack empty at the end → VALID ✓
```

### What Goes Wrong: `"({[)}"`

```
char '('  → push → stack: ['(']
char '{'  → push → stack: ['(', '{']
char '['  → push → stack: ['(', '{', '[']
char ')'  → pop '[' → matches '('? NO! → INVALID ✗
```

The stack catches mismatches immediately.

````python
def is_valid_parentheses(s: str, trace: bool = False) -> bool:
    matching = {')': '(', '}': '{', ']': '['}
    stack = []

    for ch in s:
        if ch in matching.values():
            stack.append(ch)
            if trace:
                print(f"  '{ch}' → push     → stack: {stack}")
        elif ch in matching:
            if not stack or stack[-1] != matching[ch]:
                if trace:
                    top = stack[-1] if stack else 'EMPTY'
                    print(f"  '{ch}' → pop '{top}' → mismatch! INVALID")
                return False
            popped = stack.pop()
            if trace:
                print(f"  '{ch}' → pop '{popped}' → match ✓  → stack: {stack}")

    valid = len(stack) == 0
    if trace:
        print(f"  End → stack {'empty' if valid else 'NOT empty'} → {'VALID ✓' if valid else 'INVALID ✗'}")
    return valid


print("=== Trace: '({[]})' ===")
is_valid_parentheses("({[]})", trace=True)

print("\n=== Trace: '({[)}' ===")
is_valid_parentheses("({[)}", trace=True)

print("\n=== Test Suite ===")
tests = [
    ("()", True),
    ("()[]{}", True),
    ("(]", False),
    ("({[]})", True),
    ("", True),
    ("(((", False),
    (")(", False),
    ("{[]}", True),
]
for s, expected in tests:
    result = is_valid_parentheses(s)
    status = "✓" if result == expected else "✗ FAIL"
    print(f"  {status}  is_valid({s!r:12s}) = {result}")
````

## Core Problem 2: Min Stack

**LeetCode 155** — Design a stack that supports push, pop, top, and retrieving the minimum element, all in O(1).

### The Challenge

A normal stack gives you O(1) push/pop/peek. But finding the minimum would require scanning the entire stack — O(n).

### The Trick: A Parallel "Min Stack"

Keep a second stack that tracks the current minimum at each level. Every time you push, also push the current minimum onto the min stack.

```
Operations:     push(5)    push(3)    push(7)    push(2)    push(8)

Main Stack:     [5]        [5,3]      [5,3,7]    [5,3,7,2]  [5,3,7,2,8]
Min Stack:      [5]        [5,3]      [5,3,3]    [5,3,3,2]  [5,3,3,2,2]
                                          ↑           ↑          ↑
                                       min is 3    min is 2   min is 2
                                      (7 > 3, so   (2 < 3,   (8 > 2, so
                                       keep 3)     new min!)   keep 2)

getMin() = top of min_stack → always O(1)!
```

### What Happens on Pop?

```
Before pop:   Main [5,3,7,2,8]    Min [5,3,3,2,2]
Pop:          Main [5,3,7,2]      Min [5,3,3,2]      ← pop both!
getMin() = 2  ✓

Pop again:    Main [5,3,7]        Min [5,3,3]
getMin() = 3  ✓  (2 is gone, min reverts to 3)
```

The min stack "remembers" what the minimum was before each push.

````python
class MinStack:
    def __init__(self):
        self._stack = []
        self._min_stack = []

    def push(self, val: int):
        self._stack.append(val)
        current_min = min(val, self._min_stack[-1]) if self._min_stack else val
        self._min_stack.append(current_min)

    def pop(self):
        self._stack.pop()
        self._min_stack.pop()

    def top(self) -> int:
        return self._stack[-1]

    def getMin(self) -> int:
        return self._min_stack[-1]


ms = MinStack()
operations = [5, 3, 7, 2, 8]

print("=== Pushing ===")
for val in operations:
    ms.push(val)
    print(f"  push({val}):  main={ms._stack}  min_stack={ms._min_stack}  getMin()={ms.getMin()}")

print("\n=== Popping ===")
for _ in range(3):
    top_val = ms.top()
    ms.pop()
    min_val = ms.getMin() if ms._stack else 'N/A'
    print(f"  pop({top_val}): main={ms._stack}  min_stack={ms._min_stack}  getMin()={min_val}")
````

## Core Problem 3: Next Greater Element (Monotonic Stack)

**LeetCode 496 / 503** — For each element in an array, find the next element to its right that is strictly greater. If none exists, return -1.

```
Input:  [4, 5, 2, 25, 7]
Output: [5, 25, 25, -1, -1]

  4 → next greater is 5
  5 → next greater is 25
  2 → next greater is 25
 25 → nothing greater to the right → -1
  7 → nothing greater to the right → -1
```

### Naive Approach: O(n²)

For each element, scan right until you find something bigger. Two nested loops.

### Smart Approach: Monotonic Stack — O(n)

**Idea**: Scan RIGHT to LEFT. Maintain a stack of "candidates" that could be the next greater element for future (leftward) elements.

**Key insight**: If we're processing element `x` and the stack top is ≤ `x`, that candidate is USELESS — it can never be the "next greater" for anything to the left of `x`, because `x` is closer and bigger. Pop it!

### Full Trace: `[4, 5, 2, 25, 7]`

```
Scan right to left. Stack holds candidates (bigger elements to the right).

i=4, val=7:   stack=[]        → empty, no greater → answer[4]=-1
              push 7          → stack=[7]

i=3, val=25:  stack=[7]       → top 7 ≤ 25, pop (useless)
              stack=[]        → empty, no greater → answer[3]=-1
              push 25         → stack=[25]

i=2, val=2:   stack=[25]      → top 25 > 2 → answer[2]=25
              push 2          → stack=[25, 2]

i=1, val=5:   stack=[25, 2]   → top 2 ≤ 5, pop (useless)
              stack=[25]      → top 25 > 5 → answer[1]=25
              push 5          → stack=[25, 5]

i=0, val=4:   stack=[25, 5]   → top 5 > 4 → answer[0]=5
              push 4          → stack=[25, 5, 4]

Result: [5, 25, 25, -1, -1] ✓
```

### Why O(n)?

Each element is pushed once and popped at most once. Total operations across ALL iterations = 2n → O(n).

### The "Monotonic" Part

The stack is always in **decreasing** order from bottom to top. That's why it's called a **monotonic stack**.

```
  stack bottom → [25, 5, 4] ← stack top
                  ↓   ↓  ↓
                 big      small
```

````python
def next_greater_element(nums: list[int], trace: bool = False) -> list[int]:
    n = len(nums)
    result = [-1] * n
    stack = []

    for i in range(n - 1, -1, -1):
        if trace:
            print(f"  i={i}, val={nums[i]}: stack={stack}", end="")

        while stack and stack[-1] <= nums[i]:
            popped = stack.pop()
            if trace:
                print(f" → pop {popped}", end="")

        if stack:
            result[i] = stack[-1]

        if trace:
            print(f" → answer[{i}]={result[i]}")

        stack.append(nums[i])

    return result


print("=== Trace ===")
nums = [4, 5, 2, 25, 7]
result = next_greater_element(nums, trace=True)
print(f"\nInput:  {nums}")
print(f"Output: {result}")

print("\n=== More Tests ===")
for test in [[1, 3, 2, 4], [4, 3, 2, 1], [1, 2, 3, 4], [2, 7, 3, 5, 4, 6, 8]]:
    print(f"  {test} → {next_greater_element(test)}")
````

## Core Problem 4: Evaluate Reverse Polish Notation

**LeetCode 150** — Evaluate an expression in Reverse Polish Notation (postfix).

### What Is Postfix Notation?

Normal math (infix): `(2 + 1) * 3`  
Postfix (RPN):       `2 1 + 3 *`

No parentheses needed! The order of operators defines the precedence.

### Algorithm

Scan left to right:
- **Number?** Push to stack.
- **Operator?** Pop two operands, compute, push result back.

### Trace: `["2", "1", "+", "3", "*"]` = `(2 + 1) * 3 = 9`

```
token '2'  → push         → stack: [2]
token '1'  → push         → stack: [2, 1]
token '+'  → pop 1, pop 2 → 2 + 1 = 3 → push → stack: [3]
token '3'  → push         → stack: [3, 3]
token '*'  → pop 3, pop 3 → 3 * 3 = 9 → push → stack: [9]

Final answer = 9 ✓
```

**Important**: The first popped value is the RIGHT operand, the second is the LEFT. This matters for subtraction and division:
```
  "6 2 /"  →  pop 2, pop 6  →  6 / 2 = 3  (not 2 / 6!)
```

````python
def eval_rpn(tokens: list[str], trace: bool = False) -> int:
    stack = []
    ops = {
        '+': lambda a, b: a + b,
        '-': lambda a, b: a - b,
        '*': lambda a, b: a * b,
        '/': lambda a, b: int(a / b),  # truncate toward zero
    }

    for token in tokens:
        if token in ops:
            right = stack.pop()
            left = stack.pop()
            result = ops[token](left, right)
            stack.append(result)
            if trace:
                print(f"  '{token}' → pop {right}, pop {left} → {left} {token} {right} = {result} → stack: {stack}")
        else:
            stack.append(int(token))
            if trace:
                print(f"  '{token}' → push → stack: {stack}")

    return stack[0]


print("=== Trace: (2 + 1) * 3 ===")
tokens = ["2", "1", "+", "3", "*"]
print(f"  Tokens: {tokens}")
result = eval_rpn(tokens, trace=True)
print(f"  Result: {result}")

print("\n=== Trace: 4 + 13 / 5 ===")
tokens2 = ["4", "13", "5", "/", "+"]
print(f"  Tokens: {tokens2}")
result2 = eval_rpn(tokens2, trace=True)
print(f"  Result: {result2}")

print("\n=== Trace: (10 - 6) * (8 + 2) / 4 ===")
tokens3 = ["10", "6", "-", "8", "2", "+", "*", "4", "/"]
print(f"  Tokens: {tokens3}")
result3 = eval_rpn(tokens3, trace=True)
print(f"  Result: {result3}")
````

## Core Problem 5: Queue Using Two Stacks

**LeetCode 232** — Implement a FIFO queue using only two stacks.

### The Idea

A stack is LIFO. A queue is FIFO. If you pour one stack into another, the order reverses — turning LIFO into FIFO!

```
Enqueue 1, 2, 3:

  in_stack:  [1, 2, 3]      ← always push here
  out_stack: []

Dequeue (want 1 — the first one added):

  out_stack is empty, so POUR in_stack → out_stack:

  in_stack:  [1, 2, 3]         in_stack:  []
  out_stack: []          →     out_stack: [3, 2, 1]
              pop & push                    ↑ now pop → 1!

Dequeue again (want 2):

  out_stack is NOT empty → just pop → 2!
  in_stack:  []
  out_stack: [3]
```

### Why Amortized O(1)?

Each element is moved from `in_stack` to `out_stack` exactly once. So across n operations, total moves = n → amortized O(1) per operation.

````python
class MyQueue:
    def __init__(self):
        self._in_stack = []
        self._out_stack = []

    def enqueue(self, val):
        self._in_stack.append(val)

    def _transfer(self):
        if not self._out_stack:
            while self._in_stack:
                self._out_stack.append(self._in_stack.pop())

    def dequeue(self):
        self._transfer()
        if not self._out_stack:
            raise IndexError("dequeue from empty queue")
        return self._out_stack.pop()

    def peek(self):
        self._transfer()
        if not self._out_stack:
            raise IndexError("peek at empty queue")
        return self._out_stack[-1]

    def is_empty(self):
        return not self._in_stack and not self._out_stack

    def _state(self):
        return f"in={self._in_stack}  out={self._out_stack}"


q = MyQueue()
print("=== Enqueue 1, 2, 3 ===")
for v in [1, 2, 3]:
    q.enqueue(v)
    print(f"  enqueue({v}): {q._state()}")

print("\n=== Dequeue (triggers pour) ===")
val = q.dequeue()
print(f"  dequeue() → {val}: {q._state()}")

print("\n=== Enqueue 4 (goes to in_stack) ===")
q.enqueue(4)
print(f"  enqueue(4): {q._state()}")

print("\n=== Dequeue remaining ===")
while not q.is_empty():
    val = q.dequeue()
    print(f"  dequeue() → {val}: {q._state()}")
````

## Core Problem 6: Daily Temperatures (Monotonic Stack)

**LeetCode 739** — Given daily temperatures, for each day, find how many days you have to wait until a warmer temperature. If no warmer day exists, put 0.

```
temps:  [73, 74, 75, 71, 69, 72, 76, 73]
answer: [ 1,  1,  4,  2,  1,  1,  0,  0]

Day 0 (73°): next warmer = Day 1 (74°) → wait 1 day
Day 1 (74°): next warmer = Day 2 (75°) → wait 1 day
Day 2 (75°): next warmer = Day 6 (76°) → wait 4 days
Day 3 (71°): next warmer = Day 5 (72°) → wait 2 days
...
Day 6 (76°): no warmer day → 0
Day 7 (73°): no warmer day → 0
```

### Approach: Monotonic Stack of Indices

Scan LEFT to RIGHT. The stack holds **indices** of days we haven't found a warmer day for yet.

When we see a temperature warmer than the top of the stack, we've found the answer for that stacked day!

```
i=0, temp=73: stack empty → push 0.                    stack=[0]
i=1, temp=74: 74 > temps[0]=73 → pop 0, ans[0]=1-0=1. stack=[1]
i=2, temp=75: 75 > temps[1]=74 → pop 1, ans[1]=2-1=1. stack=[2]
i=3, temp=71: 71 < temps[2]=75 → push 3.              stack=[2,3]
i=4, temp=69: 69 < temps[3]=71 → push 4.              stack=[2,3,4]
i=5, temp=72: 72 > temps[4]=69 → pop 4, ans[4]=5-4=1.
              72 > temps[3]=71 → pop 3, ans[3]=5-3=2.  stack=[2,5]
i=6, temp=76: 76 > temps[5]=72 → pop 5, ans[5]=6-5=1.
              76 > temps[2]=75 → pop 2, ans[2]=6-2=4.  stack=[6]
i=7, temp=73: 73 < temps[6]=76 → push 7.              stack=[6,7]

Remaining in stack (days 6,7) → answer stays 0.
```

````python
def daily_temperatures(temps: list[int], trace: bool = False) -> list[int]:
    n = len(temps)
    answer = [0] * n
    stack = []  # stack of indices

    for i in range(n):
        while stack and temps[i] > temps[stack[-1]]:
            prev_day = stack.pop()
            answer[prev_day] = i - prev_day
            if trace:
                print(f"  Day {i} ({temps[i]}°) is warmer than Day {prev_day} ({temps[prev_day]}°) → wait {answer[prev_day]} days")

        stack.append(i)
        if trace:
            print(f"  Push Day {i} ({temps[i]}°) → stack indices: {stack}")

    return answer


temps = [73, 74, 75, 71, 69, 72, 76, 73]
print("=== Trace ===")
result = daily_temperatures(temps, trace=True)
print(f"\nTemps:  {temps}")
print(f"Answer: {result}")
````

## Core Problem 7: Largest Rectangle in Histogram

**LeetCode 84** — Given an array of bar heights, find the area of the largest rectangle that fits inside the histogram.

```
Heights: [2, 1, 5, 6, 2, 3]

              ┌──┐
           ┌──┤  │
           │  │  │      ┌──┐
           │  │  │  ┌──┤  │
  ┌──┐     │  │  │  │  │  │
  │  │  ┌──┤  │  │  │  │  │
  └──┴──┴──┴──┴──┴──┴──┘
   2   1   5   6   2   3

Largest rectangle: height 5, width 2 → area = 10
(spans bars of height 5 and 6 — both are ≥ 5)
```

### Why Is This Hard?

For each bar, you need to know: how far LEFT and how far RIGHT can it extend? A bar of height `h` can extend in both directions as long as neighboring bars are ≥ `h`.

### Stack Approach

Maintain a stack of bar indices in **increasing height order**. When we encounter a bar shorter than the stack top, we know the tall bar can't extend further right — time to calculate its area.

```
For each bar at index i:
  While stack top is taller than heights[i]:
    Pop the top bar — it can't extend further right.
    Its right boundary = i
    Its left boundary  = new stack top (or -1 if empty)
    width = right - left - 1
    area  = height × width
  Push i onto stack.
```

### Trace: `[2, 1, 5, 6, 2, 3]`

```
i=0, h=2: push.                              stack=[0]
i=1, h=1: h[0]=2 > 1 → pop 0.
           height=2, left=-1, right=1
           width=1-(-1)-1=1, area=2×1=2.     stack=[1]
i=2, h=5: push.                              stack=[1,2]
i=3, h=6: push.                              stack=[1,2,3]
i=4, h=2: h[3]=6 > 2 → pop 3.
           height=6, left=2, right=4
           width=4-2-1=1, area=6×1=6.
          h[2]=5 > 2 → pop 2.
           height=5, left=1, right=4
           width=4-1-1=2, area=5×2=10. ★     stack=[1,4]
i=5, h=3: push.                              stack=[1,4,5]

Drain remaining:
  pop 5: h=3, left=4, right=6, w=1, area=3
  pop 4: h=2, left=1, right=6, w=4, area=8
  pop 1: h=1, left=-1, right=6, w=6, area=6

Maximum area = 10 ★
```

````python
def largest_rectangle_histogram(heights: list[int], trace: bool = False) -> int:
    stack = []  # indices of bars in increasing height order
    max_area = 0
    n = len(heights)

    def _calc_area(right_boundary):
        nonlocal max_area
        idx = stack.pop()
        height = heights[idx]
        left_boundary = stack[-1] if stack else -1
        width = right_boundary - left_boundary - 1
        area = height * width
        max_area = max(max_area, area)
        if trace:
            star = " ★" if area == max_area else ""
            print(f"    pop idx={idx}: h={height}, left={left_boundary}, right={right_boundary}, "
                  f"w={width}, area={area}{star}")

    for i in range(n):
        while stack and heights[stack[-1]] > heights[i]:
            _calc_area(i)
        stack.append(i)
        if trace:
            print(f"  i={i}, h={heights[i]}: stack={stack}")

    if trace:
        print("  Draining remaining stack:")
    while stack:
        _calc_area(n)

    return max_area


heights = [2, 1, 5, 6, 2, 3]
print("=== Trace ===")
result = largest_rectangle_histogram(heights, trace=True)
print(f"\nHeights: {heights}")
print(f"Largest rectangle area: {result}")

print("\n=== More Tests ===")
for h in [[2, 4], [1, 1, 1, 1], [6, 2, 5, 4, 5, 1, 6]]:
    print(f"  {h} → {largest_rectangle_histogram(h)}")
````

## Core Problem 8: Sliding Window Maximum

**LeetCode 239** — Given an array and window size `k`, return the maximum value in each window as it slides from left to right.

```
nums = [1, 3, -1, -3, 5, 3, 6, 7],  k = 3

Window position              Max
─────────────────────────    ───
[1  3  -1] -3  5  3  6  7    3
 1 [3  -1  -3] 5  3  6  7    3
 1  3 [-1  -3  5] 3  6  7    5
 1  3  -1 [-3  5  3] 6  7    5
 1  3  -1  -3 [5  3  6] 7    6
 1  3  -1  -3  5 [3  6  7]   7

Output: [3, 3, 5, 5, 6, 7]
```

### Naive: O(nk) — scan each window for the max.

### Smart: Monotonic Deque — O(n)

Use a **deque** that stores indices, maintaining elements in **decreasing** order.

```
Invariant: deque front is always the index of the maximum in the current window.

For each new element nums[i]:
  1. Remove from BACK: while deque back ≤ nums[i], pop it
     (those elements can never be the max — nums[i] is bigger AND newer)
  2. Add i to BACK of deque
  3. Remove from FRONT: if front index is outside the window (< i-k+1), pop it
  4. Deque front = index of current window max
```

### Trace: `[1, 3, -1, -3, 5, 3, 6, 7]`, k=3

```
i=0 (1):  deque=[0]           (just 1)
i=1 (3):  pop 0 (1≤3). deque=[1]       (3)
i=2 (-1): deque=[1,2]         window full → max=nums[1]=3
i=3 (-3): deque=[1,2,3]       front=1, still in window → max=3
i=4 (5):  pop 3,2,1 (all≤5). deque=[4]  → max=5
i=5 (3):  deque=[4,5]         → max=nums[4]=5
i=6 (6):  pop 5 (3≤6). deque=[4,6]  front=4, in window? i-k+1=4 ✓ → max=6
           Wait: pop 4 too? No, nums[4]=5 < 6 so pop it. deque=[6] → max=6
i=7 (7):  pop 6 (6≤7). deque=[7]  → max=7
```

````python
from collections import deque as Deque


def max_sliding_window(nums: list[int], k: int, trace: bool = False) -> list[int]:
    dq = Deque()  # stores indices, values in decreasing order
    result = []

    for i in range(len(nums)):
        # Remove elements smaller than current from back
        while dq and nums[dq[-1]] <= nums[i]:
            dq.pop()

        dq.append(i)

        # Remove front if outside window
        if dq[0] < i - k + 1:
            dq.popleft()

        if i >= k - 1:
            result.append(nums[dq[0]])
            if trace:
                window = nums[i - k + 1:i + 1]
                print(f"  i={i}, window={window}, deque_indices={list(dq)}, "
                      f"deque_vals={[nums[j] for j in dq]} → max={nums[dq[0]]}")

    return result


nums = [1, 3, -1, -3, 5, 3, 6, 7]
k = 3
print("=== Trace ===")
result = max_sliding_window(nums, k, trace=True)
print(f"\nInput:  {nums}, k={k}")
print(f"Output: {result}")

print("\n=== More Tests ===")
print(f"  [1] k=1 → {max_sliding_window([1], 1)}")
print(f"  [1,-1] k=1 → {max_sliding_window([1, -1], 1)}")
print(f"  [9,11] k=2 → {max_sliding_window([9, 11], 2)}")
print(f"  [4,3,2,1] k=2 → {max_sliding_window([4, 3, 2, 1], 2)}")
````

## Pattern Cheat Sheet

```
STACK & QUEUE PATTERN CHEAT SHEET:

"Matching brackets/pairs"     → Stack: push open, pop & match close
"Track running min/max"       → Two stacks (main + auxiliary)
"Next greater/smaller"        → Monotonic stack (scan right to left)
"Evaluate expression"         → Stack: push operands, pop on operator
"Process in order (BFS)"      → Queue
"Sliding window max/min"      → Monotonic deque
"Undo / backtrack"            → Stack
"Convert queue ↔ stack"       → Use two of the other
```

### Quick Decision Guide

```
             Does order matter?
              /            \
           YES              NO → probably not a stack/queue problem
            |
     Most recent first?     Arrival order?
        (LIFO)                (FIFO)
          |                     |
        STACK                 QUEUE
          |                     |
     Need min/max?       Sliding window?
      → Two stacks       → Monotonic deque
```
