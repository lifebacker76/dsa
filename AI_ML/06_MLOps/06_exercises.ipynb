# 06 — MLOps Exercises

10 hands-on exercises covering model deployment, experiment tracking, pipelines, and monitoring. No solutions provided — work through them using the companion notebooks as reference.

---
### Exercise 1 · Save and Load with joblib

Train a `GradientBoostingClassifier` on the Iris dataset. Save the model with `joblib.dump()`. In a separate cell, load it back with `joblib.load()` and verify that predictions on the test set match the original model.

---
### Exercise 2 · Save and Load a PyTorch Model

Define a simple feedforward neural network with 2 hidden layers. Train it on any small dataset (or just random data). Save the `state_dict` with `torch.save()`. In a new cell, instantiate a fresh model, load the weights back, and confirm that inference produces the same output.

---
### Exercise 3 · FastAPI Prediction Endpoint

Write a complete `app.py` (using `%%writefile`) that:
1. Loads a saved sklearn model at startup
2. Defines a Pydantic model for input validation
3. Has a `POST /predict` endpoint that accepts JSON features and returns a prediction
4. Has a `GET /health` endpoint

Use any dataset/model you like. Include a cell that shows how to test it with `requests.post()`.

---
### Exercise 4 · Dockerfile for an ML API

Write a complete `Dockerfile` (using `%%writefile`) that:
- Starts from `python:3.11-slim`
- Installs dependencies from a `requirements.txt`
- Copies the app code and model file
- Exposes port 8000
- Runs the FastAPI app with uvicorn

Also write the corresponding `requirements.txt`.

---
### Exercise 5 · Track Experiments with MLflow

Using the Wine dataset, train 3 different models (e.g., Random Forest with different `n_estimators` values). For each run:
- Log all hyperparameters with `mlflow.log_param()`
- Log accuracy and F1 score with `mlflow.log_metric()`
- Log the trained model with `mlflow.sklearn.log_model()`

After all runs, use `mlflow.search_runs()` to find the best model by accuracy.

### Exercise 6 · Complete sklearn Pipeline

Build a `Pipeline` with a `ColumnTransformer` that:
- Imputes missing numeric values with the median and scales them
- Imputes missing categorical values with the most frequent and one-hot encodes them
- Feeds everything into a classifier

Use the Titanic dataset (or any dataset with mixed types). Evaluate with cross-validation and print the classification report.

### Exercise 7 · End-to-End Training Script

Write a standalone Python script (`%%writefile train_exercise.py`) that:
1. Loads a dataset
2. Splits into train/test
3. Builds and fits a pipeline
4. Evaluates on the test set
5. Saves the pipeline to disk
6. Prints accuracy to stdout

The script should be runnable with `python train_exercise.py` with no arguments.

### Exercise 8 · Input Validation with Pydantic

Create a Pydantic `BaseModel` for a house price prediction API that validates:
- `sqft` must be a positive float between 100 and 50,000
- `bedrooms` must be an integer between 0 and 20
- `zip_code` must be a 5-digit string
- `year_built` must be an integer between 1800 and the current year
- `property_type` must be one of: "house", "condo", "townhouse"

Test it with valid and invalid inputs, and show the error messages for bad data.

### Exercise 9 · Model Monitoring Alert

Write a function `monitor_model(weekly_accuracies, threshold, window)` that:
1. Checks if the most recent accuracy is below `threshold`
2. Checks if the rolling average over the last `window` weeks shows a downward trend
3. Uses a KS test (`scipy.stats.ks_2samp`) to compare a "training distribution" vs a "recent distribution" for data drift
4. Returns a list of alert messages (empty list = all healthy)

Test it with sample data that triggers each alert.

### Exercise 10 · GitHub Actions for ML

Write a GitHub Actions YAML workflow (print it as a string or use `%%writefile`) that:
1. Triggers on push to `main`
2. Sets up Python 3.11
3. Installs dependencies from `requirements.txt`
4. Runs `pytest` on your test files
