# 05 · Computer Vision Exercises

Ten hands-on problems covering image basics, CNNs, transfer learning, and object detection.

---
### Exercise 1 — RGB & Grayscale

Load any image from CIFAR-10. Display it in full RGB and as a grayscale image side by side.  
Print the shape of each.

---
### Exercise 2 — Five Augmentations

Pick a single image from CIFAR-10. Apply **5 different augmentations** using `torchvision.transforms`  
(e.g., flip, rotate, color jitter, crop, erasing). Display all 6 images (original + 5 augmented) in a row.

---
### Exercise 3 — MNIST CNN (>98% Accuracy)

Build a CNN from scratch in PyTorch to classify MNIST digits.  
Train it and **achieve >98% accuracy** on the test set.  
Print the final test accuracy.

---
### Exercise 4 — CIFAR-10 CNN + Loss Curves

Build a deeper CNN for CIFAR-10 (color images, 10 classes).  
Train for at least 10 epochs. **Plot training loss and validation loss** on the same graph.

---
### Exercise 5 — Transfer Learning with ResNet

Load a **pretrained ResNet18** from torchvision. Replace the final fully-connected layer  
for a smaller number of classes (e.g., 10 CIFAR-10 classes). Fine-tune on CIFAR-10  
for a few epochs and report accuracy. Compare to training from scratch.

---
### Exercise 6 — Visualize Feature Maps

Take a trained CNN (from Exercise 3 or 4). Pass a single image through it and  
**extract the output of the first convolutional layer**. Display the feature maps  
as a grid of grayscale images. What patterns do you see?

---
### Exercise 7 — IoU Calculation

Implement an `iou(box1, box2)` function where each box is `[x1, y1, x2, y2]`.  
Test it with:
- Two perfectly overlapping boxes → IoU = 1.0
- Two non-overlapping boxes → IoU = 0.0
- Two partially overlapping boxes → print the IoU

Draw the two partially overlapping boxes on a blank image to visualize.

---
### Exercise 8 — Pretrained Object Detection

Load a pretrained **Faster R-CNN** from `torchvision.models.detection`.  
Run it on an image (create a synthetic scene or use a CIFAR image upscaled).  
Draw the predicted bounding boxes and class labels on the image.

---
### Exercise 9 — Augmentation Impact

Train the **same CNN architecture** on CIFAR-10 twice:
1. **Without** data augmentation (just ToTensor + Normalize)
2. **With** data augmentation (RandomCrop, RandomHorizontalFlip, ColorJitter, etc.)

Compare test accuracy after the same number of epochs. Which wins?

---
### Exercise 10 — Freeze vs Unfreeze

Using a pretrained ResNet18 on a small dataset:
1. **Freeze** all base layers → only train the final FC layer
2. **Unfreeze** all layers → fine-tune the entire network

Train both for the same number of epochs.  
Compare accuracy and training time. When would you prefer each approach?
